# 👤 Nettoyage complet des données - Clients

Ce notebook traite le dataset des clients (`olist_customers_dataset.csv`).


## 1. Configuration des chemins et du fichier

Définition des sous-dossiers et fichiers de données.


```python
from pathlib import Path
import pandas as pd
import numpy as np
import seaborn as sns
import os

RAW_DIR = Path(r"C:\Users\malek\OneDrive\Desktop\ESPRIT\3ème IA\PROJET BI\data\raw")
PROCESSED_DIR = RAW_DIR.parent / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

FILE_NAME = "olist_customers_dataset.csv"
input_path = RAW_DIR / FILE_NAME
output_path = PROCESSED_DIR / f"{input_path.stem}_cleaned.csv"
```

## 2. Vérification et lecture du fichier

Chargement et affichage des statistiques d'en-tête.


```python
if not input_path.exists():
    raise FileNotFoundError(f"Fichier introuvable : {input_path}")

df = pd.read_csv(input_path, encoding="utf-8-sig")

print("Dimensions initiales :", df.shape)
print("\nColonnes initiales :")
print(df.columns.tolist())
print("\nTypes initiaux :")
print(df.dtypes)
```

## 3. Nettoyage des noms de colonnes

Formatage des en-têtes.


```python
df.columns = df.columns.str.replace(r"[^\w]", "", regex=True)

print("\nColonnes après nettoyage des noms :")
print(df.columns.tolist())
```

## 4. Suppression des colonnes non conservées

Retrait des champs `customer_unique_id` et `customer_city`.


```python
columns_to_remove = [
    "customer_unique_id",
    "customer_city",
]

df = df.drop(columns=columns_to_remove)

print("\nColonnes restantes :")
print(df.columns.tolist())
```

## 5. Nettoyage des colonnes sélectionnées

Sanitisation des colonnes `customer_id`, `zip_code_prefix` et `state`.


```python
colonnes_texte = [
    "customer_id",
    "customer_zip_code_prefix",
    "customer_state",
]

for col in colonnes_texte:
    df[col] = df[col].str.strip()
    df[col] = df[col].replace("", pd.NA)
    df[col] = df[col].str.replace(r"[^\w]", "", regex=True)
```

## 6. Suppression des lignes/colonnes vides et des doublons

Suppression des enregistrements vides et déduplication.


```python
df = df.dropna(axis=0, how="all")
df = df.dropna(axis=1, how="all")

duplicates = df.duplicated().sum()
df = df.drop_duplicates()
```

## 7. Bilan final du nettoyage

Aperçu statistique final.


```python
print("\nNettoyage terminé.")
print("Doublons supprimés :", duplicates)
print("\nValeurs manquantes par colonne :")
print(df.isna().sum())
print("\nDimensions finales :", df.shape)
```

## 8. Export du fichier nettoyé

Écriture dans le répertoire `processed`.


```python
df.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig",
    na_rep="NULL",
)

print("\nCopie enregistrée :", output_path)
print("Le fichier original reste intact.")
```
