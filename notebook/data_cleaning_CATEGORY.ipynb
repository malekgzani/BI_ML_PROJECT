# 📦 Nettoyage complet des données - Catégories de Produits

Ce notebook contient l'intégralité du pipeline de nettoyage pour le fichier de traduction des catégories de produits (`product_category_name_translation.csv`).


## 1. Configuration des chemins et création du dossier de destination

Définition dynamique des répertoires source (`raw`) et cible (`processed`) avec la bibliothèque `pathlib`.


```python
from pathlib import Path
import pandas as pd
import numpy as np
import seaborn as sns
import os

# Configuration des chemins
RAW_DIR = Path(r"C:\Users\malek\OneDrive\Desktop\ESPRIT\3ème IA\PROJET BI\data\raw")
PROCESSED_DIR = RAW_DIR.parent / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

FILE_NAME = "product_category_name_translation.csv"
input_path = RAW_DIR / FILE_NAME
output_path = PROCESSED_DIR / f"{input_path.stem}_cleaned.csv"
```

## 2. Vérification de l'existence et lecture du fichier

Chargement du dataset avec encodage `utf-8-sig` et affichage des dimensions, colonnes et types initiaux.


```python
if not input_path.exists():
    raise FileNotFoundError(f"Fichier introuvable : {input_path}")

df = pd.read_csv(input_path, encoding="utf-8-sig")

print("Dimensions initiales :", df.shape)
print("\nColonnes initiales :")
print(df.columns.tolist())
print("\nTypes initiaux :")
print(df.dtypes)
```

## 3. Nettoyage des noms de colonnes

Normalisation des en-têtes en supprimant tout caractère non-alphanumérique.


```python
df.columns = df.columns.str.replace(r"[^\w]", "", regex=True)

print("\nColonnes après nettoyage des noms :")
print(df.columns.tolist())
```

## 4. Nettoyage des colonnes textuelles sélectionnées

Suppression des espaces inutiles (`strip`), remplacement des chaînes vides par `pd.NA` et élimination des caractères spéciaux dans le texte.


```python
colonnes_texte = [
    "product_category_name",
    "product_category_name_english",
]

for col in colonnes_texte:
    df[col] = df[col].str.strip()
    df[col] = df[col].replace("", pd.NA)
    df[col] = df[col].str.replace(r"[^\w]", "", regex=True)
```

## 5. Suppression des doublons exacts

Détection, comptage et élimination des lignes parfaitement identiques.


```python
duplicates = df.duplicated().sum()
df = df.drop_duplicates()
```

## 6. Bilan final du nettoyage

Vérification des valeurs manquantes et contrôle des dimensions finales avant export.


```python
print("\nNettoyage terminé.")
print("Doublons supprimés :", duplicates)
print("\nValeurs manquantes par colonne :")
print(df.isna().sum())
print("\nDimensions finales :", df.shape)
```

## 7. Export du fichier nettoyé

Sauvegarde du dataset prêt à l'emploi avec représentation des valeurs nulles sous forme `'NULL'`.


```python
df.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig",
    na_rep="NULL",
)

print("\nCopie enregistrée :", output_path)
print("Le fichier original reste intact.")
```
